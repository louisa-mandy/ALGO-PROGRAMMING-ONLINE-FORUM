##Exercise 1.1

A basketball player makes a penalty shot and throws a ball at 7 m/s and at an angle 60°above the horizontal. The distance from the free throw line to the hoop in basketball is 4.6 meters, the standard height of a basketball hoop 3.05 meters from the ground. The height at which the ball was thrown is 2.2 meters. Visualize the trajectory of the ball. Plot a graph with the following parameters:
-	highlight the ground line;
-	draw the trajectory with a green dashed line;
-	the trajectory must be extended until the ground line;
-	represent the hoop on the graph as you like (circle, oval, bars, etc.);
-	label the axis as “Distance” and “Height”.


```python
# Given values
v0 = 7  # velocity in m/s
angle = 60  # launch angle in degrees
g = 9.81  # acceleration due to gravity in m/s^2
x_hoop = 4.6  # horizontal distance to the hoop in meters
y_hoop = 3.05  # hoop height in meters
y0 = 2.2  # initial height in meters

# Convert angle to radians
angle_rad = np.radians(angle)

# Initial velocity components
v0x = v0 * np.cos(angle_rad)
v0y = v0 * np.sin(angle_rad)

# Time of flight (solving quadratic equation for y = 0)
t_flight = (v0y + np.sqrt(v0y**2 + 2 * g * y0)) / g

# Time values
t = np.linspace(0, t_flight, num=100)

# Compute trajectory
x = v0x * t
y = y0 + v0y * t - 0.5 * g * t**2

# Plot trajectory
plt.figure(figsize=(8, 7))
plt.plot(x, y, 'g--', label="Ball Trajectory")

# Draw ground line
plt.axhline(0, color='black', linewidth=1)

# Represent hoop as a circle
hoop_radius = 0.15
hoop_circle = plt.Circle((x_hoop, y_hoop), hoop_radius, color='orange', fill=True, label="Hoop")
plt.gca().add_patch(hoop_circle)

# Labels and title
plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")
plt.title("Basketball Shot Trajectory")
plt.legend()
plt.grid()
plt.xlim(0, max(x))
plt.ylim(0, max(y) + 0.5)

# Show plot
plt.show()
```

##Exercise 1.3

Given two functions f(x) = x2+x-1 and g(x) = -3x -1. Plot two functions and highlight the intersection points with coordinates (x, y) as the labels on them. Set the right boundaries for your graph so that the intersection points are visible. Also, draw the lines with different colors.

```python
def f(x):
    return x**2 + x - 1

def g(x):
    return -3*x - 1

x = np.linspace(-10, 10, 500)

y_f = f(x)
y_g = g(x)

x_intersections = np.array([0, -4])
y_intersections = f(x_intersections)

plt.figure(figsize=(8, 6))

plt.plot(x, y_f, label=r'$f(x) = x^2 + x - 1$', color='blue')
plt.plot(x, y_g, label=r'$g(x) = -3x - 1$', color='red')

for (xi, yi) in zip(x_intersections, y_intersections):
    plt.scatter(xi, yi, color='black')
    plt.text(xi, yi, f'({xi}, {yi})', fontsize=12, ha='right', color='black')

plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend()
plt.xlim(-5, 5)
plt.ylim(-15, 15)
plt.grid(True)
plt.show()
```

##Exercise 1.4

Given the distribution of students’ score as in the table below, draw a histogram (bar-chart) with 5 evenly spaced bins. Hint: Group the scores into 5 evenly spaced intervals of 20 points.

```python
scores = [76, 10, 21, 61, 40, 65, 40, 95, 77, 92, 59, 81, 76, 98, 57, 51, 92, 35, 17, 61, 35]

# Define number of bins
bins = np.linspace(0, 100, 6)  # 5 bins of 20 points each

# Plot histogram
plt.figure(figsize=(6, 5))
plt.hist(scores, bins=bins, edgecolor='black', alpha=0.7, color='green')

# Labels and title
plt.xlabel("Score Intervals")
plt.ylabel("Frequency")
plt.title("Histogram of Students' Scores")
plt.xticks(bins)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show plot
plt.show()
```
